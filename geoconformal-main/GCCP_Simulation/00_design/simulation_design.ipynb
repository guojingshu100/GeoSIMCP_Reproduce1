# GCCP S1--S4 simulation generator

This notebook is a short entry point for `gccp_sim_s1_s4_v1`. It generates and validates data only; it does not run any method comparison.

```python
from pathlib import Path
import sys

candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next(
    path for path in candidates
    if (path / "simulation_pilot_20260908" / "generator").is_dir()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from simulation_pilot_20260908.generator import run_pipeline

OUTPUT_ROOT = PROJECT_ROOT / "simulation_pilot_20260908"
report = run_pipeline(OUTPUT_ROOT, overwrite=True)
```

```python
import pandas as pd

display(pd.DataFrame([report["split_counts"]]))
display(pd.DataFrame(report["sigma_true_ranges"]).T.rename_axis("scenario"))
display(pd.DataFrame([
    {"check": name, "status": result["status"]}
    for name, result in report["checks"].items()
]))
print("Overall validation:", report["overall_status"])
```
